```python
import os

print("Lokasi dataset:", dataset.location)

for folder in ["train", "valid", "test"]:
    images_path = os.path.join(dataset.location, folder, "images")
    labels_path = os.path.join(dataset.location, folder, "labels")

    images = os.listdir(images_path) if os.path.exists(images_path) else []
    labels = os.listdir(labels_path) if os.path.exists(labels_path) else []

    print(f"\n{folder.upper()}")
    print("Jumlah gambar :", len(images))
    print("Jumlah anotasi:", len(labels))

print("\nISI DATA.YAML")
with open(os.path.join(dataset.location, "data.yaml"), "r") as f:
    print(f.read())
```

Import Library dan Train Model

```python
# 1. Install library YOLOv8
!pip install ultralytics

# 2. Import dan mulai Training
from ultralytics import YOLO

# Load model pretrained YOLOv8 versi nano (ringan & cepat)
model = YOLO('yolov8n.pt')

# Training model pakai dataset rock paper scissors
results = model.train(
    data=f"{dataset.location}/data.yaml",
    epochs=15,
    imgsz=640
)
```
